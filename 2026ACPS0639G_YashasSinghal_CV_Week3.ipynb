Q3. Generate a new variable with the grayscaled verion of the image and display it.

```python
gray = cv2.imread("/content/Open_CV_Img.jpg", cv2.IMREAD_GRAYSCALE)
cv2_imshow(gray)
```

Q4. Generate a new variable and store a blurred version of the image,  re-size it to (300x300) and then rotate it by 45 degrees.

```python
blurred = cv2.GaussianBlur(gray, (5, 5), 0)
resized = cv2.resize(blurred, (300, 300))
h, w = resized.shape[:2]
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(resized, M, (w, h))
cv2_imshow(rotated)
```

Q5. Draw contours on the original image using cv2.

```python
gry = cv2.cvtColor(iron_man, cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(
 thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
)
output = gry.copy()
cv2.drawContours(output, contours, -1, (0, 255, 0), 2)
cv2_imshow(output)

```

Q6. Write a text message to us on the original image and display it.

B2. Record a 10 second video of our campus lawns (be creative), and display the video through cv2.

Also, calculate the frame rate of your display and overlay it on the top left corner of the display.

```python
import time

cap = cv2.VideoCapture('/content/Grass_Vid.mp4')

previous_time = time.time()

while cap.isOpened():
  ret, frame = cap.read()
  if not ret:
    break
  current_time = time.time()
  fps = 1 / (current_time - previous_time)
  previous_time = current_time
  cv2.putText(frame, f'FPS: {fps:.1f}', (10, 30),
    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)
  cv2_imshow(frame)

cap.release()
```

B3. Use the same video that you used above, and try changing the green grass to red color.

Again, calculate the frame rate of your display and overlay it on the top left corner of the display.

```python
import numpy as np
lower_green = np.array([35, 50, 50])
upper_green = np.array([85, 255, 255])

cap = cv2.VideoCapture('Grass_Vid.mp4')

while cap.isOpened():
  ret, frame = cap.read()
  if not ret: break

  hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
  mask = cv2.inRange(hsv, lower_green, upper_green)
  cv2.putText(frame, 'GO ERC!!!!', (10, 30),
    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

  frame[mask == 255] = [0, 0, 255]
  cv2_imshow(frame)

cap.release()
```
