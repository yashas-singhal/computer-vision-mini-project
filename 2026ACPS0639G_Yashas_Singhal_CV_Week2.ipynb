Create a matrix M={$m_{ij}$} of shape (5,5) such that:
* $m_{ij}$=5i-2j if i~=~j
* $m_{ij}$=3i if i=j

Also find its inverse and its determinant.

```python
# I am considering that the first condition is for the case when i and j are not equal.

M = np.zeros((5, 5))
for i in range(5):
  for j in range(5):
    if i == j:
      M[i, j] = 3*(i+1) # I have put i+1 here as the indexing starts from 0, but I want to show my i and j from 1 to 5.
    else:
      M[i, j] = 5 * i - 2 * j

print('Matrix:', M)
Det = np.linalg.det(M)
print('Determinant:', Det)
try:
  print(np.linalg.inv(M))
except np.linalg.LinAlgError:
  print('not invertible')


```

#Matplotlib

Plot y = 2x + 3 for x = {0, 1, 2, 3, 4, 5}


```python
x = [0, 1, 2, 3, 4, 5]
y = [2*i + 3 for i in x]
plt.plot(x, y, marker='o')
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = 2x + 3')
plt.grid(True)
plt.show()
```

 Plot y = $x^2$ and y = $x^3$ on the same graph for x in range [-5,5]

 Use different colors for different functions and add a legend in the graph

```python
x = list(range(-5, 6))
y1 = [i**2 for i in x]
y2 = [i**3 for i in x]
plt.plot(x, y1, label='y = $x^2$', color='b', marker='o')
plt.plot(x, y2, label='y = $x^3$', color='g', marker='o')
plt.xlabel('X'); plt.ylabel('Y')
plt.title('y = $x^2$ and y = $x^3$')
plt.grid(True)
plt.legend()
plt.show()
```

Line Plot: Create x values from 0 to 2π with 100 points. Plot y = sin(x) and y = cos(x) on the same graph with:


* Different colors and line styles
* Proper labels and legend
* Grid enabled
* Title: "Trigonometric Functions"

```python
x = np.linspace(0,2*np.pi,100)
y_sin = np.sin(x)
y_cos = np.cos(x)
plt.plot(x, y_sin, color='b', linestyle='-', label='sin(x)')
plt.plot(x, y_cos, color='r', linestyle='--', label='cos(x)')
plt.xlabel('x values')
plt.ylabel('y values')
plt.legend()
plt.grid(True)
plt.title('Trigonometric Functions')
plt.show()
```
