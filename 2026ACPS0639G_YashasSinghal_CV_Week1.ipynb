1. Write a python program to calculate the area and perimeter of a triangle. Take side lengths as user input.

````python
import math
a = float(input("The first side of triangle is: "))
b = float(input("The second side of triangle is: "))
c = float(input("The third side of triangle is: "))

if a+b<=c or b+c<=a or a+c<=b:
    print("Such a triangle does not exist.")

else:
    p = a+b+c
    print("The perimeter of this triangle is:", p)
    s = (a+b+c)/2
    e = s*(s-a)*(s-b)*(s-c)
    A = math.sqrt(e)
    print("The area of this triangle is:", A)

# First I verified that the triangle is valid, then applied perimeter and area formulae
````

### 2.  Rock, Paper Scissors game

Create a user input-based Rock, Paper, Scissors game where the user is repeatedly prompted to enter their choice: `r` for rock, `p` for paper, or `s` for scissors. If the input is anything else, display a clear error message (e.g., “Invalid input. Please enter 'r', 'p', or 's'.") and prompt again until a valid choice is made. The game should continue until either the user or the computer reaches 5 points. After each round, display both the user's and the computer's choices, announce the winner of the round, and update the current score. Once one player reaches 5 points, declare the overall winner and end the game.


````python
user_score = 0
comp_score = 0
while user_score < 5 and comp_score < 5:

    user_input = input("Play Rock, Paper, Scissors!!!(enter 'r','p' or 's'): ")
    if user_input == "r" or user_input == "p" or user_input == "s":
        import random
        comp_num = random.randint(1,3)
        if comp_num == 1:
            comp_input = "r"
        elif comp_num == 2:
            comp_input = "p"
        elif comp_num == 3:
            comp_input = "s"


        if user_input == "r":
            user_choice = "Rock"
        elif user_input == "p":
            user_choice = "Paper"
        elif user_input == "s":
            user_choice = "Scissors"

        if comp_input == "r":
            comp_choice = "Rock"
        elif comp_input == "p":
            comp_choice = "Paper"
        elif comp_input == "s":
            comp_choice = "Scissors"

        print("You chose:", user_choice)
        print("Computer chose:", comp_choice)

        if comp_input == "r" and user_input == "s":
            result = "Lose"
        elif comp_input == "p" and user_input == "r":
            result = "Lose"
        elif comp_input == "s" and user_input == "p":
            result = "Lose"
        elif comp_input == "s" and user_input == "r":
            result = "Win"
        elif comp_input == "r" and user_input == "p":
            result = "Win"
        elif comp_input == "p" and user_input == "s":
            result = "Win"
        elif comp_input == "r" and user_input == "r":
            result = "Tie"
        elif comp_input == "p" and user_input == "p":
            result = "Tie"
        elif comp_input == "s" and user_input == "s":
            result = "Tie"

        if result == "Win":
            print("Yay!! You won this round!")
            user_score += 1

        elif result == "Lose":
            print("Ohh, you lost this round.. Let's try again")
            comp_score += 1

        else:
            print("Well, that's a tie... Let's try again!")
        print("Updated score:")
        print("Your score:", user_score)
        print("Computer score:", comp_score)

    else:
        print("Invalid input. Please enter 'r', 'p' or 's'")

if comp_score == 5:
    print("Computer won the game.")

elif user_score == 5:
    print("You won the game!!!")

# Lengthiest code... took me a lot of time, but it was fun
# Approach - first take user input and get computer input from randint, then define the cases of winning, losing and tie, and display the final winner
# Repeating this until one gets 5 points (done using while loop)
````

### 3.  Duplicates :
Write a Python program to identify and print all the duplicate characters in the string `"Electronics and Robotics Club"`

*Note:- The check should be case sensitive*

````python
text = "Electronics and Robotics Club"
counts = {}
for ch in text :
    counts [ch] = counts . get (ch,0) + 1
    duplicates = [ch for ch,c in counts.items() if c>1 and ch!=" "]
print(sorted(set(duplicates)))
# Got to know about the duplicates function for the first time
````

1. Define a function get_random(len,min=2,max=98) that:
* Generates a dataset of random numbers between `min_value` and `max_value` (both inclusive)

* The length of the dataset is determined by the parameter `len`

* Prints the dataset in the form of list,tuple and set (to show unique values)



````python
import random
def get_random(len,min =2,max =98):
    data = [random.randint(min,max) for i in range (len)]
    print("list: ", data)
    print("tuple: ", tuple (data))
    print("set: ", set(data))
# The difference in lists, tuples and sets was interesting.
````

2. Define a function **fibonacci_sequence(n, first=0, second=1)** that generates a Fibonacci-like sequence of length n.
* The sequence starts with the two numbers given by the parameters first and second (which have default values 0 and 1).

* Each subsequent number is the sum of the previous two.

* The function should return a list of length n containing the sequence.




````python
def fibonacci_sequence(n ,first=0,second=1):
    seq = [first,second]
    while len(seq)<n:
        seq.append(seq[-1]+seq[-2])
    return seq[:n]
# The logic of -1 and -2 was interesting, it makes things easier
````

### Problem Statement:

The robot should have the following:  

**Attributes:**  
- `name` → (string) name of the robot  
- `battery_level` → (integer, default = 100) battery percentage  
- `position` → (tuple, default = (0,0)) current position of the robot  

**Methods:**  
1. `move(x, y)` → Moves the robot to a new position `(x, y)` and decreases battery by 10% for each move.  
2. `charge()` → Recharges the robot’s battery back to 100%.  
3. `status()` → Prints the robot’s **name**, **battery level**, and **current position**.  

---

### Your Tasks:

2. Create at least **two Robot objects** with different names.

---

````python
class Robot:
    def __init__(self, name:str, battery_level:int=100, position:tuple=(0,0)):
        self.name = name
        self.battery_level = battery_level
        self.position = position

    def move(self,x,y):
        self.position=(x,y)
        self.battery_level -= 10

    def charge(self):
        self.battery_level = 100

    def status(self):
        return f"{self.name}: battery_level={self.battery_level} %, position={self.position}"

doraemon = Robot("Doraemon")
print(doraemon.status())
doraemon.move(5,10)
print(doraemon.status())
doraemon.move(6,7)
print(doraemon.status())
doraemon.move(6,9)
print(doraemon.status())
doraemon.move(2,4)
print(doraemon.status())
doraemon.move(3,7)
print(doraemon.status())
doraemon.move(7,3)
print(doraemon.status())
doraemon.move(4,2)
print(doraemon.status())
doraemon.move(9,6)
print(doraemon.status())
doraemon.move(7,6)
print(doraemon.status())
doraemon.move(10,5)
print(doraemon.status())
doraemon.charge()
print(doraemon.status())

dorame = Robot("Dorame")
print(dorame.status())
dorame.move(5,10)
print(dorame.status())
dorame.move(6,7)
print(dorame.status())
dorame.move(6,9)
print(dorame.status())
dorame.move(2,4)
print(dorame.status())
dorame.move(3,7)
print(dorame.status())
dorame.move(7,3)
print(dorame.status())
dorame.move(4,2)
print(dorame.status())
dorame.move(9,6)
print(dorame.status())
dorame.move(7,6)
print(dorame.status())
dorame.move(10,5)
print(dorame.status())
dorame.charge()
print(dorame.status())

# Simply followed the steps written in the question
# I could not think of any other robot names....
````
